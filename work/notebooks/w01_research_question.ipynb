## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("content_refresh_anonymized.csv")

# Keep the same usable-page conditions as the starter pipeline
usable = df[
    (df["impressions_90d"] > 0) &
    (df["content_age_days"] >= 90)
].drop_duplicates(subset="content_id")

# 1. Number of usable pages
total_pages = len(usable)

# 2. Number and percentage of declining pages
declining = usable[usable["trend_direction"] == "down"]
declining_count = len(declining)
declining_pct = declining_count / total_pages * 100

# 3. Declining pages that also have meaningful demand
declining_with_demand = usable[
    (usable["trend_direction"] == "down") &
    (usable["impressions_90d"] >= 100)
]
demand_count = len(declining_with_demand)

print(f"Usable content pages: {total_pages:,}")
print(f"Declining pages: {declining_count:,} ({declining_pct:.1f}%)")
print(f"Declining pages with at least 100 impressions: {demand_count:,}")
```
